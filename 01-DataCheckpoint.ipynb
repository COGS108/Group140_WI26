```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
import pandas as pd
import numpy as np
import os

df_raw = pd.read_csv('data/00-raw/vgsales.csv')

df_raw.head()
```

```python
# Quick Check
print("Shape:", df_raw.shape)
print("Columns:", df_raw.columns.tolist())
```

```python
# Missing Value Check

# missing counts of every columns
missing_counts = df_raw.isna().sum()

# missng percentage of every column
missing_percent = (df_raw.isna().mean() * 100).round(2)

# make a form of the above missing information
missing_df = pd.DataFrame({
    "Missing_Count": missing_counts,
    "Missing_Percent (%)": missing_percent
}).sort_values(by="Missing_Count", ascending=False)

print(missing_df)

# Check complete missing row
fully_missing_rows = df_raw.isna().all(axis=1).sum()
print(f"Completely empty rows: {fully_missing_rows}")
```

```python
# Cleaning Missing Values

print("Before cleaning:", df_raw.shape)

df_clean = df_raw.dropna(subset=["Year", "Publisher"])

print("After cleaning:", df_clean.shape)

# Confirm the result is good
print("\nRemaining missing values:")
print(df_clean.isna().sum())
```

```python
# Select Relevant Variables

selected_columns = [
    "Year",
    "Platform",
    "Genre",
    "Publisher",
    "Global_Sales"
]

df_model = df_clean[selected_columns].copy()

print("Model dataset shape:", df_model.shape)
df_model.head()
```

```python
# Year-normalized percentile

df_model["Sales_Percentile"] = (
    df_model
    .groupby("Year")["Global_Sales"]
    .rank(pct=True)
)

df_model.head()
```

```python
# Create Bestseller Levels

def classify_bestseller(p):
    if p >= 0.90:
        return "Top 10%"
    elif p >= 0.75:
        return "75–90%"
    elif p >= 0.25:
        return "25–75%"
    else:
        return "Bottom 25%"

df_model["Bestseller_Level"] = df_model["Sales_Percentile"].apply(classify_bestseller)

df_model.head()
```

```python
# Outlier Check for Global_Sales

print("Basic statistics for Global_Sales:")
print(df_model["Global_Sales"].describe())

print("\nTop 10 highest selling games:")
print(df_model.sort_values("Global_Sales", ascending=False).head(10))

print("\nBottom 10 lowest selling games:")
print(df_model.sort_values("Global_Sales", ascending=True).head(10))
```

```python
# Save Processed Dataset

os.makedirs("data/02-processed", exist_ok=True)
df_model.to_csv("data/02-processed/video_game_model_data.csv", index=False)
```

Outlier Assessment：
The distribution of Global_Sales is highly right-skewed, with a small number of extremely high-selling games (e.g., Wii Sports). However, these values reflect genuine market phenomena rather than data errors. Since our outcome variable is based on within-year percentile ranking rather than absolute sales, extreme values do not distort the modeling framework. Therefore, no trimming or transformation was applied.
