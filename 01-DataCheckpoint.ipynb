## Research Question

To what extent can pre-release structural characteristics—such as platform, genre, publisher, and release year—predict a video game’s relative commercial success, measured by its percentile rank of global sales within its release year and classified into four bestseller levels (Top 10%, 75–90%, 25–75%, Bottom 25%)?


## Background and Prior Work

Video game sales are a central indicator of commercial success, yet absolute sales figures are difficult to compare across years due to changes in market size and industry growth. To address this challenge, we study whether games can be identified as top performers based on their year-specific sales percentiles using information available prior to release.

Early empirical work on video game sales determinants provides important context for future prediction. A regression analysis by Fukuda (2017) examining factors that impact video game sales identifies key metrics such as platform, genre and review scores, which are significant predictors in explaining variation in global sales. Although the study focuses on explanatory relationships rather than prediction accuracy, it proves that systematic structure exists in game metadata and related signals, justifying the use of such characteristics for predictive models. <sup>[1](#ref1)</sup> It also highlights the challenge of working with noisy outcomes and complex interactions among various predictors, and reinforces the need for classification frameworks like yearly percentile rather than exact sales estimation.

Building on evidence that game characteristics are systematically associated with sales, more recent work explores whether these relationships can be harnessed for predictive purposes. Li et al. (2021) provides direct methodological support for our study by demonstrating that video game sales outcomes can be effectively predicted using only information available prior torelease. Their findings show that pre-release metadata and machine learning methods can be used to predict sales, indicating that a pre-release prediction framework is feasible. As stated in their paper, 'predicting sales of the video game can pre-adjust sales strategies and development plans in advance” <sup>[2](#ref2)</sup>, highlighting the practical value of pre-release sales prediction.The evidence directly supports our study and supports our main focus on the pre-sale characteristics. We extended prior work by using the same prediction framework with a year-normalized success outcome.

Besides, Kraemer et al. (2025) provides a comparison view towards our study. Their study focuses on how the post-release user andcritic reviews affect the video game sales which help us see our study’s limits in terms of predictive accuracy. In their study, they emphasize that “consumers rely heavily on online reviews to assess video game quality before a potential purchase” <sup>[3](#ref3)</sup>, highlightingthe importance of post-sale factors in sales outcome. Because these reviews’ data are unavailable before launch, our study intentionally excludes them and focuses only on the pre-release characteristics, which reflects real-world pre-launch decision-making constraints.
Additionally, with platforms like Steam increasingly supporting pre-sales, early sales may also be affected by demo quality, developer reputation, and marketing exposure.
While we acknowledge that these dynamic factors likely influence outcomes, our study focuses on prediction using consistent pre-release variables, including only factors systematically available across titles before launch and leaving more dynamic promotional elements for future research.

### Video Game Sales and Pre-Release Characteristics Dataset 



There are five primary variables used in our dataset: platform, genre, publisher, release year, and annual sales：
Platform indicates the distribution platform on which a game was released, such as Steam or PlayStation.

Genre refers to the categorical classification of each game. We adopt the genre categories provided by the dataset, which align with commonly used industry classifications, such as MOBA and platformer.
Release year ranges from 1980 to 2010, and each individual year is treated as a distinct unit in our analysis.
Annual sales represent the total global sales figure for each game within its release year.

To measure a game’s relative commercial success, we compare its annual sales against other games released in the same year and determine its percentile ranking. Based on this within-year percentile, we classify games into four bestseller levels:
Gold — Top 10%
Silver — 75th–90th percentile (Top 10%–25%)
Breakout — 25th–75th percentile
Indie Survival — Bottom 25%

This percentile-based classification allows us to evaluate commercial performance relative to market conditions within each release year.

As discussed in the Background section, our dataset includes only structured pre-release characteristics and does not incorporate dynamic factors such as marketing exposure, demo quality, developer reputation, or early user reviews. These promotional and post-release elements can have a substantial impact on sales performance, particularly on platforms such as Steam.

Because post-release factors are inherently unavailable prior to launch, they are excluded from our predictive model. Although certain pre-release promotional information may exist before launch, it is not consistently recorded or standardized within our dataset and therefore cannot be reliably incorporated.

Consequently, our model may omit influential determinants of commercial success, which could limit its overall predictive accuracy.


```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
import pandas as pd
import numpy as np
import os

df_raw = pd.read_csv('data/00-raw/vgsales.csv')

df_raw.head()
```

```python
# Quick Check
print("Shape:", df_raw.shape)
print("Columns:", df_raw.columns.tolist())
```

```python
# Missing Value Check

# missing counts of every columns
missing_counts = df_raw.isna().sum()

# missng percentage of every column
missing_percent = (df_raw.isna().mean() * 100).round(2)

# make a form of the above missing information
missing_df = pd.DataFrame({
    "Missing_Count": missing_counts,
    "Missing_Percent (%)": missing_percent
}).sort_values(by="Missing_Count", ascending=False)

print(missing_df)

# Check complete missing row
fully_missing_rows = df_raw.isna().all(axis=1).sum()
print(f"Completely empty rows: {fully_missing_rows}")
```

```python
# Cleaning Missing Values

print("Before cleaning:", df_raw.shape)

df_clean = df_raw.dropna(subset=["Year", "Publisher"])

print("After cleaning:", df_clean.shape)

# Confirm the result is good
print("\nRemaining missing values:")
print(df_clean.isna().sum())
```

```python
# Select Relevant Variables

selected_columns = [
    "Year",
    "Platform",
    "Genre",
    "Publisher",
    "Global_Sales"
]

df_model = df_clean[selected_columns].copy()

print("Model dataset shape:", df_model.shape)
df_model.head()
```

```python
# Year-normalized percentile

df_model["Sales_Percentile"] = (
    df_model
    .groupby("Year")["Global_Sales"]
    .rank(pct=True)
)

df_model.head()
```

```python
# Create Bestseller Levels

def classify_bestseller(p):
    if p >= 0.90:
        return "Top 10%"
    elif p >= 0.75:
        return "75–90%"
    elif p >= 0.25:
        return "25–75%"
    else:
        return "Bottom 25%"

df_model["Bestseller_Level"] = df_model["Sales_Percentile"].apply(classify_bestseller)

df_model.head()
```

```python
# Outlier Check for Global_Sales

print("Basic statistics for Global_Sales:")
print(df_model["Global_Sales"].describe())

print("\nTop 10 highest selling games:")
print(df_model.sort_values("Global_Sales", ascending=False).head(10))

print("\nBottom 10 lowest selling games:")
print(df_model.sort_values("Global_Sales", ascending=True).head(10))
```

```python
# Save Processed Dataset

os.makedirs("data/02-processed", exist_ok=True)
df_model.to_csv("data/02-processed/video_game_model_data.csv", index=False)
```

Outlier Assessment：
The distribution of Global_Sales is highly right-skewed, with a small number of extremely high-selling games (e.g., Wii Sports). However, these values reflect genuine market phenomena rather than data errors. Since our outcome variable is based on within-year percentile ranking rather than absolute sales, extreme values do not distort the modeling framework. Therefore, no trimming or transformation was applied.
